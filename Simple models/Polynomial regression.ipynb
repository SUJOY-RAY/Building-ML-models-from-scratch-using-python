```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([x for x in range(1000)])
y = np.array([x**2 for x in range(1000)])
```

Build design matrix (A) and vector (Y)

Each row in A is [x^2, x, 1]
Each row in B is 

```python
A = np.vstack((X**2, X, np.ones_like(X))).T
Y = y.reshape(-1, 1)

A_T = A.T
theta = np.linalg.inv(A_T @ A) @ A_T @ Y
```

```python
a, b, c = theta.flatten()
```

```python
def predict(x):
    return a * x**2 + b * x + c
```

```python
y_pred = predict(X)
```

```python
plt.scatter(X, y, color='blue', s=2, label='Actual')
plt.plot(X, y_pred, color='red', label='Predicted')
plt.xlabel('X')
plt.ylabel('y')
plt.title('Quadratic Regression (from scratch)')
plt.legend()
plt.show()
```

```python
val = 400
result = predict(val)
print(f"Predicted value at x = {val}: {result}")
```

```python
print(f"Learned coefficients: a = {a}, b = {b}, c = {c}")
```
